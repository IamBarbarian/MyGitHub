```python
# !pip install prettytable
import openpyxl as pyx
import yfinance as yf
import pandas as pd
from datetime import datetime
import numpy as np
import os as ops
import re
import sys
from prettytable import PrettyTable

class bcolors:
    HEADER = '\033[95m'
    OKBLUE = '\033[94m'
    OKCYAN = '\033[96m'
    OKGREEN = '\033[92m'
    WARNING = '\033[93m'
    FAIL = '\033[91m'
    ENDC = '\033[0m'
    BOLD = '\033[1m'
    UNDERLINE = '\033[4m'

def get_stock_price(ticker):
    try:
        data = yf.Ticker(ticker)
        hist = data.history(period="1d", interval="1d", start=None, end=None, actions=False, auto_adjust=False, back_adjust=False)
        return hist.iloc[0]["Close"]
    except Exception as ex:
        return None

table = PrettyTable()
table.field_names = ["Ticker", "Buy Price", "Current Price", "Profit"]

watch_list = {'6642.TWO':73.6, '2610.TW':25.75, '6910.TWO':32.6, '6944.TWO':388, '7812.TWO':80}

display_list = {'6642.TWO':'富致-6642','6910.TWO':'德鴻-6910','2610.TW':'華航-2610','6944.TWO':'兆聯-6944','7812.TWO':'稜研-7812'}

buy_list = ['6642.TWO','2610.TW']

benefit = 0
invent = 0

for ticker in watch_list:
    data = get_stock_price(ticker)
    stock_name = display_list[ticker]
    if ticker in buy_list:
        benefit += round((data - watch_list[ticker])*1000)
        invent += watch_list[ticker]*1000
        stock_name = "*" + display_list[ticker]
    table.add_row([stock_name, watch_list[ticker], round(data,2), round((data - watch_list[ticker])*1000)])
table.align["Ticker"] = 'l' # align with left
table.align["Buy Price"] = 'l'
table.align["Current Price"] = 'l'
table.align["Profit"] = 'l'

print(table)
print(f'Invent = {round(invent):,}')
if benefit > 0:
    print(f'{bcolors.ENDC}Profit = {bcolors.FAIL}{benefit}')
    print(f'{bcolors.ENDC}ratio = {round((benefit/invent)*100,2)}%')
else:
    print(f'{bcolors.ENDC}Loss = {bcolors.OKGREEN}{benefit}')
    print(f'{bcolors.ENDC}ratio = {bcolors.OKGREEN}{round((benefit/invent)*100,2)}%')
```
